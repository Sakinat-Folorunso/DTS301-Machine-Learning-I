# 2. Conventional programming versus machine learning

In conventional programming, we explicitly provide:

\[
\text{Rules} + \text{Data} \rightarrow \text{Output}
\]

In machine learning, we provide examples and an objective, and an algorithm learns a model:

\[
\text{Data} + \text{Examples/Targets} \rightarrow \text{Learned Model}
\]

The learned model can then be used on new data:

\[
\text{New Data} + \text{Learned Model} \rightarrow \text{Prediction}
\]

### The key idea

Machine learning is concerned with methods that learn useful patterns or relationships from data and use what was learned to make predictions or discover structure.

# 3. Machine-learning vocabulary

These terms will appear throughout DTS 301.

| Term | Meaning |
|---|---|
| **Observation / instance** | One individual example or record |
| **Variable** | A measurable or recorded property |
| **Feature** | An input variable used by a model |
| **Target / label** | The outcome a supervised model is asked to predict |
| **Dataset** | Collection of observations |
| **Model** | A mathematical/computational representation learned from data |
| **Training** | Process of learning model parameters from data |
| **Prediction** | Model output for an observation |
| **Generalization** | Ability to perform well on unseen data |

### Example

Suppose we have:

| Study hours | Attendance | Test score | Final result |
|---:|---:|---:|---|
| 5 | 80 | 65 | Pass |
| 2 | 55 | 42 | Fail |
| 7 | 90 | 78 | Pass |

Each row is an **observation**.

The first three columns are possible **features**.

`Final result` is the **target**.

# 5. The Iris dataset: our first ML dataset

We will use the classic Iris dataset to make the abstract ideas concrete.

Each observation describes an iris flower using four measurements:

- sepal length;
- sepal width;
- petal length;
- petal width.

The target identifies the flower species.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)

X = iris.data
y = iris.target

df = iris.frame

print("Dataset shape:", df.shape)
display(df.head())
````

# 6. Dataset representation

A machine-learning dataset can be represented mathematically as a matrix.

If there are \(n\) observations and \(p\) features:

\[
X \in \mathbb{R}^{n \times p}
\]

For Iris:

\[
X \in \mathbb{R}^{150 \times 4}
\]

Each observation can be represented as a feature vector:

\[
x_i =
\begin{bmatrix}
x_{i1}\\
x_{i2}\\
\vdots\\
x_{ip}
\end{bmatrix}
\]

For example, the first iris observation is:

````python
first_observation = X.iloc[0].to_numpy()

print(first_observation)
print("Number of features:", len(first_observation))
````

The feature vector is a numerical representation of one observation.

This is the bridge between the **data analysis skills from Week 0** and the **mathematical representation of machine learning** that we will develop further in Week 2.

# 9. Unsupervised learning

In supervised learning, we have an input \(X\) and a target \(y\).

In unsupervised learning, we generally have the input data \(X\) without a supplied target.

The goal may be to discover:

- groups;
- patterns;
- structure;
- lower-dimensional representations;
- unusual observations.

For example:

> "Can we group customers according to similarities in their behaviour?"

There may be no pre-existing label saying which group each customer belongs to.

We will study clustering and dimensionality reduction later in the course.

# 10. Supervised learning mathematically

Suppose our dataset is:

\[
D=\{(x_i,y_i)\}_{i=1}^{n}
\]

where:

- \(x_i\) is the feature vector for observation \(i\);
- \(y_i\) is the corresponding target.

A supervised-learning algorithm attempts to learn a function:

\[
\hat{y}=f_{\theta}(x)
\]

where:

- \(x\) = input features;
- \(y\) = actual target;
- \(\hat{y}\) = predicted target;
- \(f_{\theta}\) = learned model;
- \(\theta\) = model parameters.

### Plain-language interpretation

We want the computer to learn a useful relationship between the inputs and the target so that it can make predictions for new observations.

# 11. Prediction and loss

Suppose the true examination score is:

\[
y=70
\]

Two models make these predictions:

- Model A: \(\hat y=64\)
- Model B: \(\hat y=69\)

One simple loss function is squared error:

\[
L(y,\hat y)=(y-\hat y)^2
\]

````python
actual = 70

prediction_A = 64
prediction_B = 69

loss_A = (actual - prediction_A) ** 2
loss_B = (actual - prediction_B) ** 2

print("Model A loss:", loss_A)
print("Model B loss:", loss_B)
````

### Think

Which prediction is closer to the true value?

Which model has the smaller squared error?

The purpose of a loss function is to quantify how different a prediction is from the actual target.

Different machine-learning problems use different loss functions. We will return to this idea when studying specific algorithms.

# 12. Empirical risk

For a dataset containing \(n\) observations, we can calculate the average loss:

\[
\hat R(\theta)
=
\frac{1}{n}
\sum_{i=1}^{n}
L(f_{\theta}(x_i),y_i)
\]

This is one way to express **empirical risk**.

The broad learning idea is:

> Find model parameters that perform well according to an appropriate objective on the training data, while still aiming for good performance on unseen data.

This last part is crucial.

# 13. Training versus generalization

A model can perform extremely well on examples it has already seen.

That does not automatically mean it will perform well on new observations.

### Training performance

How well does the model perform on data used during learning?

### Generalization

How well does the model perform on data it did not see during training?

The real goal is not simply:

> "Memorize the training data."

The goal is:

> **Learn patterns that remain useful for new observations.**

## A simple analogy

Imagine a student memorizes the exact answers to 20 practice questions.

The student scores 100% when those exact questions appear again.

But on a new examination with different questions, the student performs poorly.

That is not strong evidence of learning.

Machine learning faces a similar challenge.

A model that memorizes training examples may fail to generalize.

# 14. Training, validation and test data

A common workflow separates data into different roles:

```text
Dataset
   |
   +---- Training set
   |       ↓
   |    Learn model
   |
   +---- Validation set
   |       ↓
   |    Tune/select
   |
   +---- Test set
           ↓
      Final evaluation
```

The exact workflow can vary depending on the dataset, algorithm and experimental design.

The key principle is:

> **Evaluation data should provide an honest estimate of performance on unseen data.**

We will study training, validation, testing and generalization in greater depth later.

# 16. Data leakage

Data leakage occurs when information that would not legitimately be available at prediction time enters the modelling process.

### Example

Suppose we want to predict whether a student will graduate **before graduation**.

Potential features:

- first-year GPA;
- attendance;
- assignment performance;
- number of completed courses.

Now suppose we include:

- final graduation status.

That is not a legitimate predictive feature because it is the outcome itself.

### Key question

Whenever you consider a feature, ask:

> **Would this information genuinely be available at the moment the prediction is supposed to be made?**

If not, we may have leakage.

# 17. Responsible machine learning

Machine learning systems are built from data.

Therefore, technical performance is not the only issue.

Ask:

### Data

- Who collected it?
- How was it collected?
- Who is represented?
- Who is missing?

### Target

- Is the target clearly defined?
- Is it measured reliably?
- Does it represent what we actually care about?

### Features

- Are the features available at prediction time?
- Could they contain leakage?
- Are they appropriate for the task?

### Consequences

- What happens when the model is wrong?
- Who is affected?
- How will predictions be used?

These questions will become increasingly important as we move from introductory concepts to real-world ML systems.
