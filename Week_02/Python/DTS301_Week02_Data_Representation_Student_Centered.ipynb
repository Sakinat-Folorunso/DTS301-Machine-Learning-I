# 1. From a real-world observation to data

Consider a simple student dataset:

| Student | Study Hours | Attendance | Previous Score |
|---|---:|---:|---:|
| A | 6 | 85 | 68 |
| B | 8 | 92 | 75 |
| C | 4 | 70 | 55 |

For Student A, the numerical information can be represented as:

**x = (6, 85, 68)**

This is a **feature vector**.

Each number represents one feature:

- 6 → study hours
- 85 → attendance
- 68 → previous score

### Think before running code

If we use three features for each student, how many columns should the feature matrix have?


````python
# Import NumPy, the main Python library used here for numerical arrays and matrix operations.
import numpy as np

# Create one feature vector for Student A using study hours, attendance, and previous score.
x = np.array([6, 85, 68])

# Display the feature vector.
print(x)

# Display the shape so we can see how many values it contains.
print("Shape:", x.shape)
````

### Observe

The output should show three values and a shape of `(3,)`.

That means the vector contains **3 features**.

A feature vector is therefore one numerical representation of one observation.


# 2. Feature vectors

For an observation i with p features, we can write:

**xᵢ = (xᵢ₁, xᵢ₂, …, xᵢₚ)**

For Student A:

**x = (6, 85, 68)**

For machine learning, the important idea is that an observation is converted into a numerical representation that mathematical operations can act on.

### Iris example

The classic Iris dataset has four input features:

1. sepal length
2. sepal width
3. petal length
4. petal width

One flower can therefore be represented by a four-dimensional feature vector.


````python
# Create a hypothetical Iris feature vector containing four measurements.
iris_x = np.array([5.1, 3.5, 1.4, 0.2])

# Display the feature vector.
print(iris_x)

# Display the number of features represented by the vector.
print("Number of features:", iris_x.shape[0])
````

# 3. Feature matrices

When we have many observations, we place their feature vectors together to form a **feature matrix**.

For n observations and p features:

**X ∈ ℝⁿˣᵖ**

Rows represent observations.  
Columns represent features.

For the student dataset:

| Student | Study Hours | Attendance | Previous Score |
|---|---:|---:|---:|
| A | 6 | 85 | 68 |
| B | 8 | 92 | 75 |
| C | 4 | 70 | 55 |

there are:

- 3 observations
- 3 features

Therefore:

**X ∈ ℝ³ˣ³**


````python
# Create a feature matrix for the three students.
X = np.array([
    [6, 85, 68],
    [8, 92, 75],
    [4, 70, 55]
])

# Display the feature matrix.
print(X)

# Display the matrix dimensions.
print("Shape:", X.shape)

# Display the number of observations.
print("Observations:", X.shape[0])

# Display the number of features.
print("Features:", X.shape[1])
````

### Check your understanding

If a dataset contains **500 observations and 10 features**, what should the shape of its feature matrix be?

**Answer:** `500 × 10`

In mathematical notation:

**X ∈ ℝ⁵⁰⁰ˣ¹⁰**


# 4. Inspecting rows and columns

Rows correspond to observations.

Columns correspond to features.

Let's inspect the student matrix.


````python
# Select the first row, representing Student A.
student_A = X[0]

# Display Student A's feature vector.
print("Student A:", student_A)

# Select the first column, representing study hours.
study_hours = X[:, 0]

# Display the study-hours feature for all students.
print("Study hours:", study_hours)

# Select the second column, representing attendance.
attendance = X[:, 1]

# Display the attendance feature for all students.
print("Attendance:", attendance)
````

# 5. Vector operations

Vectors can be manipulated mathematically.

Let:

**a = (2, 4, 6)**

**b = (1, 3, 5)**

We can perform:

- addition
- subtraction
- scalar multiplication
- dot product

These operations are important because machine learning algorithms repeatedly perform mathematical operations on numerical representations.


````python
# Create vector a.
a = np.array([2, 4, 6])

# Create vector b.
b = np.array([1, 3, 5])

# Add the two vectors element by element.
print("a + b =", a + b)

# Subtract vector b from vector a element by element.
print("a - b =", a - b)

# Multiply vector a by the scalar value 2.
print("2a =", 2 * a)

# Calculate the dot product of a and b.
print("a · b =", np.dot(a, b))
````

### Work out the dot product

For:

**a = (2, 4, 6)**

**b = (1, 3, 5)**

the dot product is:

**a · b = (2×1) + (4×3) + (6×5) = 44**

The dot product combines corresponding values and adds the results.

### Predict

Before running the next example, predict the result of:

**(3, 2, 1) · (1, 2, 3)**


````python
# Create the first prediction vector.
u = np.array([3, 2, 1])

# Create the second prediction vector.
v = np.array([1, 2, 3])

# Calculate and display the dot product.
print("u · v =", np.dot(u, v))
````

# 6. Matrix–vector multiplication

A matrix can operate on a vector.

Consider:

**A = [[1, 2], [3, 4]]**

and:

**x = (5, 6)**

The multiplication is:

**Ax = (17, 39)**

The matrix has shape `2 × 2` and the vector has length 2, so the multiplication is valid.

In Python, the `@` operator is commonly used for matrix multiplication.


````python
# Create the 2 by 2 matrix A.
A = np.array([
    [1, 2],
    [3, 4]
])

# Create the two-element vector x.
x = np.array([5, 6])

# Multiply matrix A by vector x.
Ax = A @ x

# Display the result.
print("Ax =", Ax)

# Display the shape of the result.
print("Result shape:", Ax.shape)
````

# 7. Matrix dimension compatibility

Matrix multiplication is only possible when the **inner dimensions match**.

For example:

`(3 × 2)(2 × 1)` is valid.

The result has shape:

`3 × 1`

But:

`(3 × 2)(3 × 1)` is not valid because the inner dimensions do not match.

### Rule

For:

`(m × n)(n × p)`

the result has shape:

`m × p`

This dimension rule is essential when working with machine learning data.


````python
# Create a 3 by 2 matrix.
M = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

# Create a vector with two elements so the multiplication is valid.
v2 = np.array([10, 20])

# Multiply the matrix by the compatible vector.
result = M @ v2

# Display the result.
print(result)

# Display the result shape.
print("Result shape:", result.shape)
````

# 8. Linear transformations

A linear transformation changes a vector using a matrix.

A simple example is scaling.

Let:

**A = [[2, 0], [0, 2]]**

and:

**x = (2, 3)**

Then:

**Ax = (4, 6)**

The transformation doubles both coordinates.

Linear transformations appear throughout machine learning, including feature transformations and the operations used inside many models.


````python
# Create a simple scaling matrix.
scale = np.array([
    [2, 0],
    [0, 2]
])

# Create a two-dimensional input vector.
point = np.array([2, 3])

# Apply the linear transformation.
transformed_point = scale @ point

# Display the original vector.
print("Original:", point)

# Display the transformed vector.
print("Transformed:", transformed_point)
````

# 9. A simple feature transformation

Features can have very different numerical scales.

For example:

- study hours may range from 0 to 20
- attendance may range from 0 to 100
- scores may range from 0 to 100

One simple transformation is to divide percentage-like variables by 100.

For a value x:

**x′ = x / 100**

This is only a simple demonstration. Later topics will introduce more systematic feature scaling methods.


````python
# Create a copy of the student feature matrix so the original remains unchanged.
X_scaled = X.astype(float).copy()

# Scale the attendance column from a 0–100 range to a 0–1 range.
X_scaled[:, 1] = X_scaled[:, 1] / 100

# Scale the previous-score column from a 0–100 range to a 0–1 range.
X_scaled[:, 2] = X_scaled[:, 2] / 100

# Display the transformed matrix.
print(X_scaled)
````
