# 5. Python Functions

A function is a reusable block of code that performs a task.

Functions become extremely important in data science because we repeatedly perform operations such as:

- calculating statistics;
- transforming data;
- cleaning data;
- training models;
- evaluating models.

The basic pattern is:

```text
input → function → output
```


````python
# Define a function that calculates the average of two scores.
def calculate_average(score_1, score_2):
    # Add the two scores together.
    total = score_1 + score_2

    # Divide the total by two to obtain the average.
    average = total / 2

    # Return the calculated average to the caller.
    return average

# Call the function using two scores.
result = calculate_average(70, 90)

# Display the returned result.
print("Average:", result)
````

# 6. Introduction to NumPy

**NumPy** (Numerical Python) provides efficient tools for numerical computing.

Machine learning works extensively with numerical representations. NumPy gives us structures and operations for:

- vectors;
- matrices;
- arrays;
- numerical calculations;
- vectorized operations.

A NumPy array is especially important because later in the course we will represent machine-learning data as matrices.


````python
# Import NumPy and give it the common alias np.
import numpy as np

# Create a one-dimensional NumPy array containing four numbers.
scores = np.array([70, 80, 90, 85])

# Display the NumPy array.
print(scores)

# Display the type of the object.
print(type(scores))

# Display the shape of the array.
print("Shape:", scores.shape)

# Display the number of dimensions.
print("Dimensions:", scores.ndim)
````

### Understanding `shape`

The **shape** tells us how the elements of an array are organised.

For a one-dimensional array:

```text
(4,)
```

means that there are four elements.

For a two-dimensional array:

```text
(3, 4)
```

means:

- 3 rows;
- 4 columns.

This idea will become central in Week 2.


````python
# Create a two-dimensional array with three students and four features.
X = np.array([
    [80, 75, 90, 85],
    [60, 65, 70, 68],
    [92, 88, 95, 90]
])

# Display the matrix.
print(X)

# Display the number of rows and columns.
print("Shape:", X.shape)

# Display the number of dimensions.
print("Dimensions:", X.ndim)

# Display the number of elements in the matrix.
print("Number of elements:", X.size)
````

## 7. NumPy Indexing and Slicing

A two-dimensional NumPy array can be accessed using:

```python
array[row, column]
```

As with Python lists, indexing begins at zero.

If `X` has shape `(3, 4)`:

- row indices are `0, 1, 2`;
- column indices are `0, 1, 2, 3`.


````python
# Display the first row of X.
print("First row:", X[0, :])

# Display the second row of X.
print("Second row:", X[1, :])

# Display the first column of X.
print("First column:", X[:, 0])

# Display the value in the first row and first column.
print("First element:", X[0, 0])

# Display the first two rows and first two columns.
print("Top-left block:")
print(X[:2, :2])
````

# 8. Vectorized Operations

One of NumPy's major advantages is that we can perform an operation on an entire array without writing a separate loop for every element.

For example, if scores are stored in an array, we can add 5 to every score in one operation.

This is called **vectorization**.


````python
# Create an array containing three scores.
scores = np.array([60, 70, 80])

# Add five marks to every score.
adjusted_scores = scores + 5

# Display the original scores.
print("Original:", scores)

# Display the adjusted scores.
print("Adjusted:", adjusted_scores)

# Multiply every score by two.
double_scores = scores * 2

# Display the multiplied scores.
print("Doubled:", double_scores)
````

# 9. Basic NumPy Statistics

NumPy provides common statistical operations.

We will use:

- `mean()` — average;
- `median()` — middle value;
- `min()` — smallest value;
- `max()` — largest value;
- `std()` — standard deviation;
- `sum()` — total.

These are basic building blocks for EDA.


````python
# Create an array containing examination scores.
scores = np.array([55, 60, 65, 70, 75, 80, 85, 90])

# Calculate the mean score.
mean_score = np.mean(scores)

# Calculate the median score.
median_score = np.median(scores)

# Find the minimum score.
minimum_score = np.min(scores)

# Find the maximum score.
maximum_score = np.max(scores)

# Calculate the standard deviation.
standard_deviation = np.std(scores)

# Display the calculated statistics.
print("Mean:", mean_score)
print("Median:", median_score)
print("Minimum:", minimum_score)
print("Maximum:", maximum_score)
print("Standard deviation:", standard_deviation)
````

# 10. Introduction to Pandas

**Pandas** is widely used for working with structured and tabular data.

A Pandas **DataFrame** can be thought of as a table containing:

- rows representing observations;
- columns representing variables or features.

This is one of the most common data structures used before machine learning.


````python
# Import Pandas and give it the common alias pd.
import pandas as pd

# Create a dictionary containing student information.
data = {
    "student": ["Amina", "Bola", "Chidi", "David", "Efe"],
    "attendance": [85, 70, 92, 60, 78],
    "test_score": [78, 65, 88, 55, 72],
    "final_score": [82, 68, 91, 58, 75]
}

# Convert the dictionary into a Pandas DataFrame.
df = pd.DataFrame(data)

# Display the DataFrame.
print(df)
````

# 14. Missing Data

Real datasets are rarely perfect.

Common problems include:

- missing values;
- inconsistent entries;
- duplicated observations;
- incorrect data types;
- unusual values.

We will begin with missing values.

A missing value means that information expected for an observation is absent.


````python
# Create a small DataFrame containing one missing test score.
missing_data = pd.DataFrame({
    "student": ["Amina", "Bola", "Chidi"],
    "test_score": [78, np.nan, 88],
    "final_score": [82, 68, 91]
})

# Display the DataFrame containing the missing value.
print(missing_data)

# Count missing values in each column.
print("Missing values by column:")
print(missing_data.isna().sum())
````

### Handling Missing Values

There is no single correct strategy for every dataset.

Possible approaches include:

- removing observations;
- removing variables;
- filling missing values using a justified method;
- investigating why values are missing.

The correct choice depends on the context and the modelling task.

For now, the important skill is to **detect and think about missingness before modelling**.


````python
# Create a copy of the DataFrame so that the original data remain unchanged.
filled_data = missing_data.copy()

# Calculate the mean of the available test scores.
mean_test_score = filled_data["test_score"].mean()

# Fill the missing test score using the calculated mean.
filled_data["test_score"] = filled_data["test_score"].fillna(mean_test_score)

# Display the completed DataFrame.
print(filled_data)
````

### Important Question

Is replacing a missing value with the mean **always** appropriate?

No.

Before choosing a method, ask:

- Why is the value missing?
- How much data is missing?
- Could the missingness itself contain information?
- Would the method introduce bias?
- Is the information available legitimately?

This is the beginning of responsible data analysis.

Numbers are important, but visualizations can reveal patterns that are difficult to see in tables.

We will use **Matplotlib** for basic plots.

We will explore:

- histograms;
- scatter plots;
- boxplots.

Visualization is part of EDA, not decoration.


````python
# Import Matplotlib's plotting module and give it the common alias plt.
import matplotlib.pyplot as plt

# Create a simple list of examination scores.
scores = [55, 60, 65, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

# Create a histogram of the examination scores.
plt.hist(scores, bins=5)

# Add a label to the horizontal axis.
plt.xlabel("Examination Score")

# Add a label to the vertical axis.
plt.ylabel("Number of Students")

# Add a title to the plot.
plt.title("Distribution of Examination Scores")

# Display the histogram.
plt.show()
````

# 16. Correlation

Correlation measures the strength and direction of a **linear relationship** between numerical variables.

A correlation coefficient is commonly represented by:

$$r \in [-1,1]$$

Very roughly:

- values near `+1` indicate a strong positive linear relationship;
- values near `-1` indicate a strong negative linear relationship;
- values near `0` indicate little linear relationship.

Correlation does **not** establish causation.


````python
# Calculate the correlation coefficient between study hours and examination scores.
correlation = np.corrcoef(study_hours, exam_scores)[0, 1]

# Display the correlation coefficient.
print("Correlation:", correlation)
````
